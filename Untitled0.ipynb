```python
# Ejemplo 1: Estructuras de datos (Listas y Diccionarios)
print("--- EJEMPLO 1: Gestión de Estudiantes ---")

# Diccionario que almacena información de un alumno
estudiante = {
    "nombre": "Jessica",
    "carrera": "Ciencias de la Computación",
    "calificaciones": [95, 88, 92, 90],
}

# Calcular el promedio de calificaciones
promedio = sum(estudiante["calificaciones"]) / len(
    estudiante["calificaciones"]
)

print(f"Estudiante: {estudiante['nombre']}")
print(f"Carrera: {estudiante['carrera']}")
print(f"Promedio final: {promedio:.2f}")
```

```python
# Ejemplo 2: Uso básico de la librería Pandas
import pandas as pd

print("--- EJEMPLO 2: Tabla de Datos con Pandas ---")

# Crear datos de productos de una tienda o financiera
data = {
    "Producto": ["Tarjeta de Crédito", "Préstamo Personal", "Hipoteca", "Ahorro"],
    "Tasa_Interes": [15.5, 12.0, 9.8, 4.5],
    "Activos": [1200, 3400, 8900, 5600],
}

# Convertir a un DataFrame (tabla)
df = pd.DataFrame(data)

# Mostrar la tabla y un resumen estadístico
display(df)
print("\nPromedio de Tasas de Interés:", df["Tasa_Interes"].mean())
```

```python
# Ejemplo 3: Gráfica matemática con Matplotlib
import matplotlib.pyplot as plt
import numpy as np

print("--- EJEMPLO 3: Visualización Gráfica ---")

# Generar valores para X (de -10 a 10)
x = np.linspace(-10, 10, 100)
# Calcular la función cuadrática Y = x^2
y = x**2

# Crear la gráfica
plt.figure(figsize=(6, 4))
plt.plot(x, y, color="blue", label="y = x^2")
plt.title("Gráfica de una función cuadrática")
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Ejemplo 4: Simulación de base de datos con objetos
print("--- EJEMPLO 4: Simulación de Base de Datos ---")


class RegistroFinanciero:

  def __init__(self, id_transaccion, monto, tipo):
    self.id_transaccion = id_transaccion
    self.monto = monto
    self.tipo = tipo  # 'Ingreso' o 'Egreso'


# Simulando una "tabla" con una lista de registros
base_de_datos_falsa = [
    RegistroFinanciero(1, 1500.0, "Ingreso"),
    RegistroFinanciero(2, 450.0, "Egreso"),
    RegistroFinanciero(3, 3000.0, "Ingreso"),
]

# Consultar y mostrar los registros
print(f"Total de transacciones en la base de datos: {len(base_de_datos_falsa)}")
for transaccion in base_de_datos_falsa:
  print(
      f"ID: {transaccion.id_transaccion} | Tipo: {transaccion.tipo} | Monto:"
      f" ${transaccion.monto}"
  )
```
